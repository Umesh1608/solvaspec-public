# SolvaSpec: predicting a UV-Vis spectrum for one molecule

This notebook walks through a single end-to-end query of the SolvaSpec pipeline. We use **avobenzone** (a widely-used UVA filter) in **methanol** and compare against the published experimental λ_max ≈ 358 nm.

```python
from solvaspec import predict_solvent_spectrum

result = predict_solvent_spectrum(
    smiles="CC(C)(C)c1ccc(cc1)/C(O)=C\\C(=O)c1ccc(OC)cc1",  # avobenzone, chelated enol
    solvent="methanol",
    n_conformers=8,
    return_uncertainty=True,
)
print(f"lambda_max     = {result.lambda_max_nm:.1f} nm  (experimental ~ 358 nm)")
print(f"epsilon_max    = {result.eps_max:.2e} M^-1 cm^-1")
print(f"UVA integrated = {result.uva:.3f}")
print(f"UVB integrated = {result.uvb:.3f}")
print(f"5-fold sigma   = {result.cross_fold_sigma_nm:.1f} nm  (OOD signal)")
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(result.wavelength_nm, result.absorbance, color="#2E86AB", lw=1.6)
ax.axvline(result.lambda_max_nm, color="#cc4444", ls="--", lw=0.9,
           label=fr"$\lambda_\mathrm{{max}}$ = {result.lambda_max_nm:.0f} nm")
ax.set_xlabel(r"$\lambda$ (nm)")
ax.set_ylabel(r"absorbance ($\mathrm{M}^{-1}\mathrm{cm}^{-1}$)")
ax.set_title("Avobenzone (enol) in methanol")
ax.legend(loc="upper right", frameon=False)
ax.grid(alpha=0.3)
plt.show()
```
